## Setup

```
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

df = pd.read_csv('sales_data.csv')
df['Date'] = pd.to_datetime(df['Date'])

print("✅ Data loaded!")
print(f"Shape: {df.shape}")
```

---

## Solusi Latihan 1: Line Plot Revenue per Bulan

```
# Extract bulan dan aggregate
df['Month'] = df['Date'].dt.to_period('M').astype(str)
monthly_revenue = df.groupby('Month')['Revenue'].sum()

# Plot
plt.figure(figsize=(12, 6))
plt.plot(monthly_revenue.index, monthly_revenue.values, 
         marker='o', linewidth=2, markersize=8, 
         color='steelblue', markerfacecolor='orange',
         markeredgecolor='black', markeredgewidth=1)

plt.title('Total Revenue per Bulan', fontsize=14, fontweight='bold')
plt.xlabel('Bulan', fontsize=12)
plt.ylabel('Revenue (USD)', fontsize=12)
plt.grid(True, alpha=0.3, linestyle='--')
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()

print("\n💰 Revenue per Bulan:")
print(monthly_revenue)
print(f"\n📊 Total Revenue: ${monthly_revenue.sum():,.2f}")
print(f"📈 Average Monthly Revenue: ${monthly_revenue.mean():,.2f}")
```

**Penjelasan:**
- `dt.to_period('M')` mengconvert tanggal ke periode bulan
- `groupby().sum()` menghitung total revenue per bulan
- Styling dengan marker dan warna membuat grafik lebih menarik
- Grid membantu membaca nilai dengan lebih mudah

---

## Solusi Latihan 2: Horizontal Bar Chart Quantity per Kategori

```
# Aggregate dan sort
quantity_by_category = df.groupby('Category')['Quantity'].sum().sort_values(ascending=True)

# Plot
plt.figure(figsize=(10, 6))
bars = plt.barh(quantity_by_category.index, quantity_by_category.values,
                color='coral', alpha=0.8, edgecolor='black', linewidth=1.5)

# Tambahkan nilai di ujung bar
for i, bar in enumerate(bars):
    width = bar.get_width()
    plt.text(width + 2, bar.get_y() + bar.get_height()/2,
             f'{int(width)}', ha='left', va='center', fontweight='bold')

plt.title('Total Quantity Terjual per Kategori', fontsize=14, fontweight='bold')
plt.xlabel('Total Quantity', fontsize=12)
plt.ylabel('Kategori', fontsize=12)
plt.grid(True, alpha=0.3, axis='x')
plt.tight_layout()

plt.show()

print("\n📦 Quantity per Kategori:")
print(quantity_by_category.sort_values(ascending=False))
```

**Penjelasan:**
- `ascending=True` dalam sort membuat bar paling besar di atas
- `plt.barh()` untuk horizontal bar chart
- Text annotation menampilkan nilai exact
- `edgecolor` memberikan border pada bar

## Solusi Latihan 3: Scatter Plot Age vs Revenue by Region

```
plt.figure(figsize=(12, 7))

# Color map untuk region
colors = {'North': 'red', 'South': 'blue', 'East': 'green', 'West': 'orange'}

# Plot untuk setiap region
for region in df['Region'].unique():
    subset = df[df['Region'] == region]
    plt.scatter(subset['Customer_Age'], subset['Revenue'],
               label=region, c=colors[region], alpha=0.6, s=80,
               edgecolors='black', linewidth=0.5)

plt.title('Hubungan Customer Age vs Revenue by Region', fontsize=14, fontweight='bold')
plt.xlabel('Customer Age', fontsize=12)
plt.ylabel('Revenue (USD)', fontsize=12)
plt.legend(title='Region', fontsize=10)
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.show()

# Analisis korelasi
print("\n📊 Korelasi Age vs Revenue:")
print(f"Correlation: {df['Customer_Age'].corr(df['Revenue']):.3f}")

# Stats per region
print("\n📍 Average Revenue per Region:")
print(df.groupby('Region')['Revenue'].mean().sort_values(ascending=False))
```

**Penjelasan:**
- Loop untuk setiap region memudahkan color coding
- Dictionary colors untuk konsistensi warna
- `s=80` mengatur ukuran marker
- Correlation analysis memberikan insight tambahan

## Solusi Latihan 4: Grouped Bar Chart Price by Category & Gender

```
# Aggregate data
price_by_cat_gender = df.groupby(['Category', 'Gender'])['Price'].mean().unstack(fill_value=0)

# Setup bar positions
x = np.arange(len(price_by_cat_gender))
width = 0.35

# Plot
fig, ax = plt.subplots(figsize=(12, 6))
bars1 = ax.bar(x - width/2, price_by_cat_gender['M'], width,
               label='Male', color='steelblue', alpha=0.8, edgecolor='black')
bars2 = ax.bar(x + width/2, price_by_cat_gender['F'], width,
               label='Female', color='pink', alpha=0.8, edgecolor='black')

# Labels dan styling
ax.set_xlabel('Category', fontsize=12)
ax.set_ylabel('Average Price (USD)', fontsize=12)
ax.set_title('Average Price by Category and Gender', fontsize=14, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(price_by_cat_gender.index, rotation=45)
ax.legend(title='Gender')
ax.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()

print("\n💵 Average Price by Category and Gender:")
print(price_by_cat_gender)
print("\n🔍 Insight:")
for cat in price_by_cat_gender.index:
    diff = price_by_cat_gender.loc[cat, 'M'] - price_by_cat_gender.loc[cat, 'F']
    if abs(diff) > 10:
        higher = 'Male' if diff > 0 else 'Female'
        print(f"  - {cat}: {higher} spend ${abs(diff):.2f} more on average")
```

**Penjelasan:**
- `unstack()` mengubah MultiIndex menjadi columns
- Positioned bars dengan `x - width/2` dan `x + width/2`
- Insight analysis memberikan interpretasi data

## Solusi Latihan 5: Seaborn Heatmap Revenue by Region & Category

```
# Pivot table
revenue_pivot = df.pivot_table(values='Revenue', 
                                index='Region', 
                                columns='Category', 
                                aggfunc='mean')

# Plot
plt.figure(figsize=(10, 6))
sns.heatmap(revenue_pivot, 
            annot=True,           # Show numbers
            fmt='.1f',            # Format: 1 decimal
            cmap='YlOrRd',        # Yellow-Orange-Red colormap
            linewidths=0.5,       # Lines between cells
            cbar_kws={'label': 'Average Revenue (USD)'})

plt.title('Average Revenue by Region and Category', fontsize=14, fontweight='bold')
plt.xlabel('Category', fontsize=12)
plt.ylabel('Region', fontsize=12)
plt.tight_layout()

plt.show()

print("\n💰 Average Revenue Pivot Table:")
print(revenue_pivot)
print("\n🏆 Highest Average Revenue:")
max_val = revenue_pivot.max().max()
max_loc = revenue_pivot.stack().idxmax()
print(f"  {max_loc[0]} - {max_loc[1]}: ${max_val:.2f}")
```

**Penjelasan:**
- `pivot_table` lebih fleksibel dari `groupby().unstack()`
- `fmt='.1f'` format angka dengan 1 desimal
- `YlOrRd` colormap bagus untuk nilai revenue
- Stack().idxmax() menemukan kombinasi dengan nilai tertinggi

## Solusi Latihan 6: Box Plot & Violin Plot Comparison

```
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Subplot 1: Box plot
sns.boxplot(data=df, x='Category', y='Price', ax=axes[0], palette='Set2')
axes[0].set_title('Price Distribution by Category', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Category')
axes[0].set_ylabel('Price (USD)')
axes[0].tick_params(axis='x', rotation=45)
axes[0].grid(True, alpha=0.3, axis='y')

# Subplot 2: Violin plot
sns.violinplot(data=df, x='Region', y='Customer_Age', ax=axes[1], 
               palette='muted', inner='quartile')
axes[1].set_title('Customer Age Distribution by Region', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Region')
axes[1].set_ylabel('Customer Age')
axes[1].grid(True, alpha=0.3, axis='y')

plt.suptitle('Distribution Comparison: Box Plot vs Violin Plot', 
            fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()

plt.show()

# Statistics
print("\n📊 Price Statistics by Category:")
print(df.groupby('Category')['Price'].describe()[['mean', 'std', 'min', 'max']])
print("\n👥 Age Statistics by Region:")
print(df.groupby('Region')['Customer_Age'].describe()[['mean', 'std', 'min', 'max']])
```

**Penjelasan:**
- Box plot menunjukkan quartiles dan outliers
- Violin plot menunjukkan density distribution
- `inner='quartile'` menampilkan quartile lines di violin
- Statistics summary memberikan nilai numerik

## Solusi Latihan 7: Stacked Histogram with Hue

```
plt.figure(figsize=(12, 6))

sns.histplot(data=df, x='Price', hue='Gender', 
             multiple='stack',      # Stacked bars
             kde=True,              # KDE overlay
             bins=20,               # Number of bins
             alpha=0.7,             # Transparency
             palette='Set2',        # Color palette
             edgecolor='black',     # Bar borders
             linewidth=0.5)

plt.title('Price Distribution by Gender (Stacked Histogram)', 
         fontsize=14, fontweight='bold')
plt.xlabel('Price (USD)', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.legend(title='Gender', labels=['Female', 'Male'])
plt.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()

# Analysis
print("\n💵 Price Statistics by Gender:")
price_stats = df.groupby('Gender')['Price'].describe()
print(price_stats)

print("\n📊 Price Distribution:")
print(f"Male - Mean: ${df[df['Gender']=='M']['Price'].mean():.2f}")
print(f"Female - Mean: ${df[df['Gender']=='F']['Price'].mean():.2f}")

# Statistical test
from scipy import stats
male_prices = df[df['Gender']=='M']['Price']
female_prices = df[df['Gender']=='F']['Price']
t_stat, p_value = stats.ttest_ind(male_prices, female_prices)
print(f"\n🔬 T-test: t-statistic={t_stat:.3f}, p-value={p_value:.3f}")
if p_value < 0.05:
    print("  → Significant difference in prices between genders")
else:
    print("  → No significant difference in prices between genders")
```

**Penjelasan:**
- `multiple='stack'` membuat histogram bertumpuk
- KDE overlay menunjukkan smooth distribution
- Statistical test (t-test) untuk validasi perbedaan
- Comparison analysis memberikan insight yang lebih dalam

## Solusi Latihan 8: Dashboard 2x2

```
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle('Sales Analytics Dashboard', fontsize=16, fontweight='bold')

# 1. Top-left: Revenue per bulan
monthly_rev = df.groupby(df['Date'].dt.to_period('M'))['Revenue'].sum()
axes[0, 0].plot(monthly_rev.index.astype(str), monthly_rev.values, 
                'o-', linewidth=2, markersize=8, color='steelblue')
axes[0, 0].set_title('Monthly Revenue Trend', fontweight='bold')
axes[0, 0].set_xlabel('Month')
axes[0, 0].set_ylabel('Revenue (USD)')
axes[0, 0].tick_params(axis='x', rotation=45)
axes[0, 0].grid(True, alpha=0.3)

# 2. Top-right: Pie chart kategori
category_counts = df['Category'].value_counts()
colors_pie = plt.cm.Set3(range(len(category_counts)))
axes[0, 1].pie(category_counts.values, labels=category_counts.index, 
               autopct='%1.1f%%', colors=colors_pie, startangle=90)
axes[0, 1].set_title('Transaction Distribution by Category', fontweight='bold')

# 3. Bottom-left: Scatter Price vs Quantity
categories = df['Category'].unique()
colors_scatter = plt.cm.Set2(range(len(categories)))
for i, cat in enumerate(categories):
    subset = df[df['Category'] == cat]
    axes[1, 0].scatter(subset['Price'], subset['Quantity'], 
                      label=cat, alpha=0.6, s=80, c=[colors_scatter[i]])
axes[1, 0].set_title('Price vs Quantity by Category', fontweight='bold')
axes[1, 0].set_xlabel('Price (USD)')
axes[1, 0].set_ylabel('Quantity')
axes[1, 0].legend(title='Category', loc='best')
axes[1, 0].grid(True, alpha=0.3)

# 4. Bottom-right: Top 5 products
top5_products = df.groupby('Product')['Revenue'].sum().nlargest(5).sort_values()
axes[1, 1].barh(top5_products.index, top5_products.values, 
                color='coral', alpha=0.8, edgecolor='black')
axes[1, 1].set_title('Top 5 Products by Revenue', fontweight='bold')
axes[1, 1].set_xlabel('Revenue (USD)')
for i, v in enumerate(top5_products.values):
    axes[1, 1].text(v + 20, i, f'${v:,.0f}', va='center')

plt.tight_layout()
plt.show()

print("\n📊 Dashboard Summary:")
print(f"Total Revenue: ${df['Revenue'].sum():,.2f}")
print(f"Total Transactions: {len(df)}")
print(f"Average Transaction: ${df['Revenue'].mean():,.2f}")
print(f"Best Month: {monthly_rev.idxmax()}")
print(f"Best Category: {df.groupby('Category')['Revenue'].sum().idxmax()}")
```

**Penjelasan:**
- Dashboard 2x2 memberikan overview lengkap
- Setiap subplot menceritakan story berbeda
- Kombinasi plot types memberikan perspektif yang beragam
- Summary statistics melengkapi visualisasi

## Solusi Latihan 9: Correlation Analysis

```
# 1. Correlation matrix
numerical_cols = df.select_dtypes(include=[np.number]).columns
corr_matrix = df[numerical_cols].corr()

# 2. Heatmap dengan mask
plt.figure(figsize=(10, 8))
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
sns.heatmap(corr_matrix, mask=mask, annot=True, fmt='.2f',
            cmap='coolwarm', center=0, square=True, linewidths=0.5,
            cbar_kws={'label': 'Correlation Coefficient'})
plt.title('Correlation Matrix (Lower Triangle)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# 3. Top 3 korelasi
corr_pairs = corr_matrix.unstack()
corr_pairs = corr_pairs[corr_pairs < 1.0]  # Exclude diagonal
top3_corr = corr_pairs.abs().sort_values(ascending=False).head(3)

print("\n🔝 Top 3 Korelasi Tertinggi:")
for pair, corr in top3_corr.items():
    actual_corr = corr_pairs[pair]
    print(f"  {pair[0]} vs {pair[1]}: {actual_corr:.3f}")

# 4. Scatter plots untuk top 3
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
fig.suptitle('Top 3 Strongest Correlations', fontsize=14, fontweight='bold')

for idx, (pair, corr) in enumerate(top3_corr.items()):
    actual_corr = corr_pairs[pair]
    sns.regplot(data=df, x=pair[0], y=pair[1], ax=axes[idx],
               scatter_kws={'alpha': 0.5}, line_kws={'color': 'red'})
    axes[idx].set_title(f'{pair[0]} vs {pair[1]}\nCorr: {actual_corr:.3f}', 
                        fontweight='bold')
    axes[idx].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Interpretation
print("\n🔍 Interpretation:")
for pair, corr in top3_corr.items():
    actual_corr = corr_pairs[pair]
    if abs(actual_corr) > 0.7:
        strength = "kuat"
    elif abs(actual_corr) > 0.4:
        strength = "sedang"
    else:
        strength = "lemah"
    
    direction = "positif" if actual_corr > 0 else "negatif"
    print(f"  - {pair[0]} dan {pair[1]}: Korelasi {strength} {direction}")
```

**Penjelasan:**
- Mask upper triangle menghilangkan duplikasi
- Unstack() flatten matrix untuk sorting
- Top correlations dipilih berdasarkan absolute value
- Regression line menunjukkan trend
- Interpretation memberikan makna dari angka

## Solusi Latihan 10: Advanced FacetGrid

```
# FacetGrid dengan row dan col
g = sns.FacetGrid(df, row='Gender', col='Region', 
                  height=4, aspect=1.2, 
                  margin_titles=True)

# Map histogram
g.map(sns.histplot, 'Revenue', bins=10, kde=True, color='steelblue', alpha=0.7)

# Tambahkan mean line
mean_revenue = df['Revenue'].mean()
g.map(plt.axvline, x=mean_revenue, color='red', linestyle='--', linewidth=2)

# Customize
g.set_axis_labels('Revenue (USD)', 'Count')
g.set_titles(row_template="Gender: {row_name}", col_template="Region: {col_name}")
g.fig.suptitle('Revenue Distribution by Gender and Region', 
              fontsize=16, fontweight='bold', y=1.02)

# Add legend untuk mean line
g.add_legend(label_order=['Distribution', 'Overall Mean'])

plt.show()

# Detailed analysis
print("\n📊 Revenue Statistics by Gender and Region:")
print(df.groupby(['Gender', 'Region'])['Revenue'].agg(['count', 'mean', 'std', 'min', 'max']))

print("\n🎯 Insights:")
summary = df.groupby(['Gender', 'Region'])['Revenue'].mean().unstack()
print("\nAverage Revenue Matrix:")
print(summary)

# Temukan kombinasi terbaik dan terburuk
best_combo = summary.stack().idxmax()
worst_combo = summary.stack().idxmin()
print(f"\n🏆 Best Combination: {best_combo[0]} in {best_combo[1]} - ${summary.loc[best_combo]:.2f}")
print(f"📉 Lowest Combination: {worst_combo[0]} in {worst_combo[1]} - ${summary.loc[worst_combo]:.2f}")

# Perbandingan gender per region
print("\n👥 Gender Comparison per Region:")
for region in df['Region'].unique():
    male_rev = df[(df['Region']==region) & (df['Gender']=='M')]['Revenue'].mean()
    female_rev = df[(df['Region']==region) & (df['Gender']=='F')]['Revenue'].mean()
    diff = male_rev - female_rev
    if abs(diff) > 50:
        higher = "Male" if diff > 0 else "Female"
        print(f"  {region}: {higher} revenue ${abs(diff):.2f} higher")
```

**Penjelasan:**
- FacetGrid membuat multiple plots dengan struktur terorganisir
- Row dan col parameter membuat grid layout
- Map() function menerapkan plot ke semua facets
- Mean line memberikan reference point
- Detailed analysis mengungkap patterns tersembunyi
- Best/worst combination membantu decision making
